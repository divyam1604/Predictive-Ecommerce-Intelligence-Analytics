# Predictive E-Commerce Intelligence Analytics — Corrected Pipeline

This is the hardened rebuild of the original notebook. Eleven defects are fixed;
each is tagged `[FIX n]` in the cell that addresses it.

| # | Defect in original | Consequence | Fix |
|---|---|---|---|
| 1 | `payments` and `reviews` joined one-to-many onto a line-item table | 6,493 duplicate rows, GMV inflated 5.05% (R$800k) | pre-aggregate both to order grain |
| 2 | No cardinality checks on 7 merges | fan-out went unnoticed | `validate=` + row-count assertion |
| 3 | Null categories silently dropped from charts | 1,627 rows invisible | explicit `unknown` bucket |
| 4 | `revenue = price + freight` labelled as revenue | freight is a cost pass-through, not income | GMV / freight / total split |
| 5 | `delivery_group()` returns `"Late"` for `NaN` | 3,421 cancelled + undelivered orders labelled Late | `pd.isna()` guard, `Not Delivered` bucket |
| 6 | No `order_status` filter | cancelled orders counted in revenue | recognise on delivery |
| 7 | AOV computed as mean of line items | R$140.68 instead of R$159.83 | aggregate to order grain first |
| 8 | Linear forecast, no split, no metrics ever printed | unvalidated model | clean window, walk-forward CV, naive baseline |
| 9 | K-Means unscaled, k hardcoded, `count` not `nunique` | clusters were just spend bins | RFM + log + scaling + silhouette |
| 10 | No actionable model | analysis stops at description | 1–2 star review classifier |
| 11 | Dashboard loads a 68 MB raw CSV | slow, unfilterable | pre-aggregated summary tables |

```python
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (average_precision_score, mean_absolute_error,
                             r2_score, roc_auc_score, silhouette_score)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# point this at the folder holding the olist_*.csv files
DATA_DIR = next((p for p in [Path("data"), Path("."), Path("../data")]
                 if (p / "olist_orders_dataset.csv").exists()), Path("data"))
CUR = "R$"   # [FIX 4] Olist is a Brazilian marketplace - Reais, not dollars

print("reading from:", DATA_DIR.resolve())
```

## Load

```python
names = ["customers", "orders", "order_items", "order_payments",
         "order_reviews", "products", "sellers"]
raw = {n: pd.read_csv(DATA_DIR / f"olist_{n}_dataset.csv") for n in names}
raw["translation"] = pd.read_csv(DATA_DIR / "product_category_name_translation.csv")

customers, orders, items = raw["customers"], raw["orders"], raw["order_items"]
payments, reviews = raw["order_payments"], raw["order_reviews"]
products, sellers, translation = raw["products"], raw["sellers"], raw["translation"]

for k, v in raw.items():
    print(f"  {k:16s} {v.shape}")
```

## [FIX 1] Collapse one-to-many tables to order grain *before* merging

`order_payments` has 103,886 rows for 99,441 orders (2,961 orders paid with more
than one instrument) and `order_reviews` has 99,224 rows with 547 duplicated
orders. The original notebook merged both straight onto a line-item table, which
duplicated every affected line and double-counted `price`.

**Rule: never join a one-to-many table onto a fact table without aggregating it first.**

```python
pay_order = (
    payments.sort_values("payment_value", ascending=False)
    .groupby("order_id")
    .agg(payment_value=("payment_value", "sum"),
         payment_type=("payment_type", "first"),          # highest-value instrument
         payment_installments=("payment_installments", "max"),
         n_payment_instruments=("payment_sequential", "max"))
    .reset_index()
)

reviews["review_creation_date"] = pd.to_datetime(reviews["review_creation_date"])
rev_order = (reviews.sort_values("review_creation_date")
             .drop_duplicates("order_id", keep="last")     # keep the latest review
             [["order_id", "review_score", "review_creation_date"]])

print(f"payments {len(payments):>7,} -> {len(pay_order):>7,}")
print(f"reviews  {len(reviews):>7,} -> {len(rev_order):>7,}")
assert pay_order.order_id.is_unique and rev_order.order_id.is_unique
```

## [FIX 2] Build the fact table with cardinality guards

`validate="many_to_one"` makes pandas **raise** instead of silently fanning out.
The assertion is the safety net: if the row count changes on a join that should
not change it, the build fails loudly instead of producing a plausible wrong number.

```python
n0 = len(items)
fact = (
    items
    .merge(products,    on="product_id",            how="left", validate="many_to_one")
    .merge(translation, on="product_category_name", how="left", validate="many_to_one")
    .merge(sellers,     on="seller_id",             how="left", validate="many_to_one")
    .merge(orders,      on="order_id",              how="left", validate="many_to_one")
    .merge(customers,   on="customer_id",           how="left", validate="many_to_one")
    .merge(pay_order,   on="order_id",              how="left", validate="many_to_one")
    .merge(rev_order,   on="order_id",              how="left", validate="many_to_one")
)
assert len(fact) == n0, f"fan-out detected: {n0} -> {len(fact)}"
print(f"fact table {fact.shape}   (was 119,143 x 45 - 6,493 rows were duplicates)")
```

## [FIX 3] Missing categories become an explicit bucket, not a silent drop

```python
miss = fact["product_category_name_english"].isna().sum()
fact["category"] = fact["product_category_name_english"].fillna("unknown")
print(f"{miss:,} rows had no category ({100*miss/len(fact):.2f}%) -> labelled 'unknown'")
```

## [FIX 4] Separate GMV from freight

`price + freight` is gross transacted value, not revenue. Freight is a cost
pass-through — Olist's own revenue is a commission on GMV. Keeping the three
figures separate lets you answer either question.

```python
fact["gmv"] = fact["price"]
fact["freight"] = fact["freight_value"]
fact["line_total"] = fact["price"] + fact["freight_value"]

for c in ["order_purchase_timestamp", "order_delivered_customer_date",
          "order_estimated_delivery_date", "order_approved_at"]:
    fact[c] = pd.to_datetime(fact[c])

fact["order_month"] = fact["order_purchase_timestamp"].dt.to_period("M")
print(f"freight is {100*fact.freight.sum()/fact.line_total.sum():.1f}% of transacted value")
```

## [FIX 5] The `NaN` bug in `delivery_group()`

In Python `NaN < 0` is `False` **and** `NaN == 0` is `False`, so every missing
value fell through the original function's `else` branch and was labelled
`"Late"`. Those 3,421 rows are orders that were cancelled, unavailable, or still
in transit — a completely different business condition. It inflated the late rate
from ~6.5% to 9.2%.

```python
fact["delivery_days"] = (fact["order_delivered_customer_date"]
                         - fact["order_purchase_timestamp"]).dt.days
fact["delivery_gap"] = (fact["order_delivered_customer_date"]
                        - fact["order_estimated_delivery_date"]).dt.days


def delivery_group(x):
    if pd.isna(x):
        return "Not Delivered"      # <-- the fix
    if x < 0:
        return "Early"
    if x == 0:
        return "On Time"
    return "Late"


fact["delivery_status"] = fact["delivery_gap"].apply(delivery_group)
print(fact["delivery_status"].value_counts().to_string())
```

## [FIX 6] Revenue recognises on delivery

The original counted cancelled and unavailable orders in GMV. Keep the full fact
table for funnel and cancellation analysis; use the delivered slice for money.

```python
delivered = fact[fact["order_status"] == "delivered"].copy()
print(f"delivered {len(delivered):,} of {len(fact):,} rows "
      f"(excluded {len(fact)-len(delivered):,})")
print(fact.order_status.value_counts().to_string())
```

## [FIX 7] KPIs at the correct grain

Average order value is per **order**, not per line item. The original reported the
mean of `revenue` across a fanned-out line-item table.

```python
order_level = (
    delivered.groupby("order_id")
    .agg(order_value=("line_total", "sum"),
         gmv=("gmv", "sum"),
         freight=("freight", "sum"),
         customer_unique_id=("customer_unique_id", "first"),
         customer_state=("customer_state", "first"),
         review_score=("review_score", "first"),
         delivery_days=("delivery_days", "first"),
         delivery_status=("delivery_status", "first"),
         order_month=("order_month", "first"),
         purchase_ts=("order_purchase_timestamp", "first"),
         n_items=("order_item_id", "count"))
    .reset_index()
)

kpi = {
    "Gross merchandise value": f"{CUR}{delivered.gmv.sum():,.0f}",
    "Freight collected":       f"{CUR}{delivered.freight.sum():,.0f}",
    "Total transacted":        f"{CUR}{delivered.line_total.sum():,.0f}",
    "Delivered orders":        f"{order_level.order_id.nunique():,}",
    "Unique customers":        f"{order_level.customer_unique_id.nunique():,}",
    "Average order value":     f"{CUR}{order_level.order_value.mean():,.2f}",
    "Items per order":         f"{order_level.n_items.mean():.2f}",
    "Average review":          f"{order_level.review_score.mean():.2f}",
    "1-2 star rate":           f"{(order_level.review_score <= 2).mean()*100:.1f}%",
}
for k, v in kpi.items():
    print(f"  {k:26s} {v}")
```

## EDA — same charts, corrected inputs

Note the headline finding survives every fix: delivery timeliness drives ratings,
price does not.

```python
fig, ax = plt.subplots(2, 2, figsize=(14, 9))

monthly_plot = order_level.groupby("order_month")["order_value"].sum()
monthly_plot.index = monthly_plot.index.astype(str)
monthly_plot.plot(ax=ax[0, 0], marker="o", title="Monthly revenue (delivered orders)")
ax[0, 0].set_ylabel(f"{CUR}")
ax[0, 0].tick_params(axis="x", rotation=45)

(delivered.groupby("category")["line_total"].sum().nlargest(10)
 .sort_values().plot(kind="barh", ax=ax[0, 1], title="Top 10 categories by revenue"))

sns.barplot(data=order_level, x="review_score", y="delivery_days", ax=ax[1, 0])
ax[1, 0].set_title("Delivery duration by review score")

order_level.delivery_status.value_counts().plot(
    kind="bar", ax=ax[1, 1], title="Delivery performance")
ax[1, 1].tick_params(axis="x", rotation=0)

plt.tight_layout(); plt.show()

print("\nAverage review and 1-star rate by delivery status:")
print(order_level.groupby("delivery_status")
      .agg(orders=("order_id", "size"),
           avg_review=("review_score", "mean"),
           pct_one_star=("review_score", lambda s: (s == 1).mean()*100))
      .round(2).to_string())

print("\nCorrelations:")
print(order_level[["delivery_days", "order_value", "freight", "review_score"]]
      .corr().round(3).to_string())
```

## [FIX 8] Forecast: clean window, walk-forward validation, a baseline to beat

Three things the original never did:

1. **Truncate the series.** 2016-09/10/12 are a near-empty pilot and 2018-09/10 are
   the dataset's collection cut-off, not a business collapse. Fitting through them
   is what dragged R² down to 0.276; on the clean window it is 0.82.
2. **Validate chronologically.** A random split lets the model train on month 20 and
   predict month 10. Walk-forward, one step ahead, is the only honest option.
3. **Compare against a naive baseline.** If last month's value beats your model,
   you do not have a model.

R² is deliberately **not** reported on the holdout — with six points its denominator
is unstable and the number is meaningless. MAE and MAPE are the right metrics here.

```python
monthly = (order_level.groupby("order_month")["order_value"].sum()
           .reset_index().rename(columns={"order_value": "revenue"}))
monthly["order_month"] = monthly["order_month"].astype(str)

CLEAN_START, CLEAN_END = "2017-01", "2018-08"
clean = monthly[(monthly.order_month >= CLEAN_START)
                & (monthly.order_month <= CLEAN_END)].reset_index(drop=True)
clean["t"] = np.arange(len(clean))
clean["month_no"] = pd.to_datetime(clean.order_month).dt.month
X_all = pd.concat([clean[["t"]],
                   pd.get_dummies(clean.month_no, prefix="m", drop_first=True)],
                  axis=1).astype(float)
y = clean.revenue.values
print(f"full series {len(monthly)} months -> clean window {len(clean)} "
      f"({CLEAN_START} to {CLEAN_END})")

STEPS = 6
res = {"naive (last value)": [], "linear trend": [], "trend + seasonality": []}
for i in range(len(clean) - STEPS, len(clean)):
    res["naive (last value)"].append(y[i - 1])
    res["linear trend"].append(
        LinearRegression().fit(clean[["t"]][:i], y[:i]).predict(clean[["t"]][i:i+1])[0])
    res["trend + seasonality"].append(
        LinearRegression().fit(X_all[:i], y[:i]).predict(X_all[i:i+1])[0])

truth = y[-STEPS:]
print(f"\nWalk-forward, 1-step-ahead, last {STEPS} months:")
for name, pred in res.items():
    pred = np.array(pred)
    print(f"  {name:22s} MAE={CUR}{mean_absolute_error(truth, pred):>9,.0f}   "
          f"MAPE={np.mean(np.abs((truth-pred)/truth))*100:5.1f}%")

full_fit = LinearRegression().fit(clean[["t"]], y)
print(f"\nin-sample R2 on clean window: "
      f"{r2_score(y, full_fit.predict(clean[['t']])):.3f}  "
      f"(original notebook, untrimmed: 0.276)")
print(f"fitted trend: {CUR}{full_fit.coef_[0]:,.0f}/month")
```

> **Result to carry into any conversation about this project:** the naive baseline
> beats the linear trend by roughly 3x on MAE. Revenue plateaued through 2018 while
> a straight line keeps climbing, so the trend model is not fit for forecasting —
> it is a description of 2017 growth, not a prediction. The honest next step is
> SARIMA or Prophet with a 12-month seasonal term, or admitting that 20 monthly
> points is not enough data to forecast from.

## [FIX 9] Segmentation: full RFM, distinct orders, log-transform, scaling, chosen k

Four separate problems in the original:

- **No scaling.** K-Means minimises Euclidean distance. `spending` had ~300x the
  standard deviation of `orders`, so the clusters were spend bins with the
  frequency dimension contributing nothing.
- **`count` instead of `nunique`** on `order_id` — counted fanned-out rows, not orders.
- **No Recency.** RFM without R is missing its strongest churn signal.
- **k hardcoded to 4** with no elbow or silhouette check (and the README said 3).

```python
snapshot = order_level.purchase_ts.max()
rfm = (order_level.groupby("customer_unique_id")
       .agg(recency=("purchase_ts", lambda s: (snapshot - s.max()).days),
            frequency=("order_id", "nunique"),        # distinct orders, not rows
            monetary=("order_value", "sum"))
       .reset_index())
print(f"{len(rfm):,} customers | repeat rate {(rfm.frequency > 1).mean()*100:.1f}%")

feat = rfm[["recency", "frequency", "monetary"]].copy()
feat["monetary"] = np.log1p(feat["monetary"])          # tame the right skew
Xs = StandardScaler().fit_transform(feat)

samp = np.random.default_rng(0).choice(len(Xs), 20000, replace=False)
scores = {}
for k in range(2, 7):
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(Xs)
    scores[k] = (km.inertia_, silhouette_score(Xs[samp], km.labels_[samp]))
    print(f"  k={k}  inertia={km.inertia_:>10,.0f}  silhouette={scores[k][1]:.3f}")

best_k = max(scores, key=lambda k: scores[k][1])
print(f"\nselected k={best_k} by silhouette (original hardcoded k=4)")

km = KMeans(n_clusters=best_k, random_state=42, n_init=10).fit(Xs)
rfm["segment"] = km.labels_
print(rfm.groupby("segment").agg(customers=("customer_unique_id", "size"),
                                 recency=("recency", "mean"),
                                 frequency=("frequency", "mean"),
                                 monetary=("monetary", "mean")).round(1).to_string())
```

Silhouette collapses after k=2 because **frequency is degenerate at a 3% repeat
rate** — 97% of customers bought exactly once, so there is almost nothing to
cluster on. The honest read is that RFM is the wrong framework for this business.

The practical alternative is a value tier based on *first-order* characteristics,
which is actionable on day one and does not require a purchase history:

```python
rfm["value_tier"] = pd.qcut(rfm.monetary, [0, .5, .8, .95, 1],
                            labels=["Low", "Mid", "High", "Top"])
print(rfm.groupby("value_tier", observed=True)
      .agg(customers=("customer_unique_id", "size"),
           avg_spend=("monetary", "mean"),
           pct_of_revenue=("monetary", lambda s: 100*s.sum()/rfm.monetary.sum()))
      .round(1).to_string())
```

## [FIX 10] The model the project was missing

The strongest finding in the data — delivery drives satisfaction — was only ever
described, never operationalised. A classifier that flags an order likely to
receive a 1–2 star review turns it into something an operations team can act on
*before* the review lands.

```python
md_ = order_level.dropna(subset=["review_score", "delivery_days"]).copy()
md_["late"] = (md_.delivery_status == "Late").astype(int)
md_["target"] = (md_.review_score <= 2).astype(int)
md_ = md_.merge(delivered.groupby("order_id")["category"].first(), on="order_id")

Xc = pd.DataFrame({
    "delivery_days": md_.delivery_days,
    "late": md_.late,
    "freight": md_.freight,
    "order_value": md_.order_value,
    "n_items": md_.n_items,
    "state": md_.customer_state.astype("category").cat.codes,
    "category": md_.category.astype("category").cat.codes,
})
yc = md_.target

Xtr, Xte, ytr, yte = train_test_split(Xc, yc, test_size=.25,
                                      stratify=yc, random_state=42)
clf = HistGradientBoostingClassifier(max_iter=200, random_state=42).fit(Xtr, ytr)
pr = clf.predict_proba(Xte)[:, 1]

print(f"positive class rate  {yc.mean()*100:.1f}%")
print(f"ROC-AUC              {roc_auc_score(yte, pr):.3f}")
print(f"PR-AUC               {average_precision_score(yte, pr):.3f}  "
      f"(random baseline {yte.mean():.3f})")
```

Accuracy would be a misleading metric here — always predicting "happy" scores 87%.
ROC-AUC and PR-AUC against the 12.8% base rate are the honest measures.

## [FIX 11] Export pre-aggregated tables for the dashboard

The original wrote a 68 MB CSV and the Dash app loaded all of it at import. The
dashboard only ever renders five aggregations, so ship those instead — roughly
170 KB total.

```python
out = Path("dashboard_data")
out.mkdir(exist_ok=True)

summary = {
    "monthly": monthly,
    "by_category": delivered.groupby("category").agg(
        revenue=("line_total", "sum"), orders=("order_id", "nunique"),
        avg_review=("review_score", "mean")).reset_index(),
    "by_state": order_level.groupby("customer_state").agg(
        revenue=("order_value", "sum"), orders=("order_id", "nunique"),
        aov=("order_value", "mean")).reset_index(),
    "by_seller": delivered.groupby("seller_id").agg(
        revenue=("line_total", "sum"), orders=("order_id", "nunique"),
        avg_review=("review_score", "mean")).reset_index(),
    "reviews": order_level.review_score.value_counts().sort_index()
                          .rename_axis("review_score").reset_index(name="orders"),
}
for k, v in summary.items():
    v.to_csv(out / f"summary_{k}.csv", index=False)

# slim line-item table so the dashboard can filter interactively
dash_cols = ["order_id", "order_month", "category", "customer_state", "seller_id",
             "review_score", "delivery_status", "delivery_days", "line_total"]
dash_fact = delivered[dash_cols].copy()
dash_fact["order_month"] = dash_fact["order_month"].astype(str)
dash_fact.to_csv(out / "dash_fact.csv.gz", index=False, compression="gzip")
print(f"dash_fact.csv.gz: {(out/'dash_fact.csv.gz').stat().st_size/1024/1024:.1f} MB "
      f"for {len(dash_fact):,} rows (was 68 MB / 119,143 rows x 45 cols)")

total_kb = sum(p.stat().st_size for p in out.glob("summary_*.csv")) / 1024
print(f"summary tables written: {total_kb:.0f} KB (was a 68 MB CSV)")
```

## Remaining known limitations

Fixing code does not fix data. State these before anyone asks:

- **No cost data.** GMV is not profit — no COGS, no commission rate, no returns.
  "Biggest category" and "most profitable category" may not be the same answer.
- **20 monthly points is thin for forecasting.** One seasonal cycle is not enough
  to fit seasonality reliably; that is why the naive baseline wins.
- **`geolocation` is still unused.** Real shipping distance is available and would
  likely be the strongest single feature in the review classifier.
- **Review bias.** The score distribution is J-shaped (56% five-star, 13% one-star),
  so the mean is a poor summary. Track the 1–2 star rate instead.
- **Correlation, not causation.** Delivery duration is confounded with distance,
  seller, and product type. Establishing causation needs stratification on route
  or an experiment on promise accuracy.
